### 4. Creating Financial KPIs

Profit is calculated as revenue minus budget. Return on Investment (ROI) measures revenue relative to the production budget. Both metrics will be calculated in millions of US dollars where applicable. ROI will be calculated only for movies with a known, positive budget.

```python
df["profit_musd"] = df["revenue_musd"] - df["budget_musd"]

df["roi"] = np.where(
    df["budget_musd"] > 0,
    df["revenue_musd"] / df["budget_musd"],
    np.nan
)

display(
    df[
        ["title", "budget_musd", "revenue_musd", "profit_musd", "roi"]
    ].head(10)
)
```

### 5. Defining a User-Defined Ranking Function

I will create a reusable function to rank movies by a selected KPI. This avoids repeating similar code for each ranking and allows the same function to identify both the highest and lowest values.

```python
def rank_movies(data, metric, ascending=False, top_n=5, min_votes=None):
    """
    Rank movies by a numeric metric.

    Parameters:
        data: DataFrame containing movie information
        metric: Column used for ranking
        ascending: False for highest first, True for lowest first
        top_n: Number of movies to return
        min_votes: Optional minimum vote count

    Returns:
        DataFrame of ranked movies
    """
    result = data.copy()

    result[metric] = pd.to_numeric(result[metric], errors="coerce")
    result = result.dropna(subset=[metric])

    if min_votes is not None:
        result = result[result["vote_count"] >= min_votes]

    return result.sort_values(
        metric, ascending=ascending
    ).head(top_n)
```

### 6. Ranking Movies by KPIs

I will use the ranking function to identify movies with the highest revenue, highest budget, highest and lowest profit, highest and lowest ROI, most votes, highest and lowest ratings, and greatest popularity. Rankings will exclude missing KPI values and apply the specified eligibility rules.

```python
print("\nHighest revenue movies:")
display(rank_movies(df, "revenue_musd")[
    ["title", "revenue_musd", "budget_musd"]
])
```

```python
print("\nHighest budget movies:")
display(rank_movies(df, "budget_musd")[
    ["title", "budget_musd", "revenue_musd"]
])
```

```python
print("\nHighest profit movies:")
display(rank_movies(df, "profit_musd")[
    ["title", "profit_musd", "revenue_musd", "budget_musd"]
])

print("\nLowest profit movies:")
display(rank_movies(df, "profit_musd", ascending=True)[
    ["title", "profit_musd", "revenue_musd", "budget_musd"]
])
```

```python
print("\nHighest ROI movies:")
eligible_roi = df[df["budget_musd"] >= 10].copy()

display(rank_movies(eligible_roi, "roi")[
    ["title", "budget_musd", "revenue_musd", "roi"]
])
print("\nLowest ROI movies:")
display(rank_movies(eligible_roi, "roi", ascending=True)[
    ["title", "budget_musd", "revenue_musd", "roi"]
])
```

```python
print("\nHighest vote count movies:")
display(rank_movies(df, "vote_count")[
    ["title", "vote_count", "vote_average"]
])
```

```python
print("\nHighest rated movies (with at least 10 votes):")
display(rank_movies(df, "vote_average", min_votes=10)[
    ["title", "vote_average", "vote_count"]
])

print("\nLowest rated movies (with at least 10 votes):")
display(rank_movies(
    df, "vote_average", ascending=True, min_votes=10
)[["title", "vote_average", "vote_count"]])
```

```python
print("\nHighest popularity movies:")
display(rank_movies(df, "popularity")[
    ["title", "popularity", "vote_average"]
])
```

### 7. Advanced Movie Filtering and Search Queries

I will filter the dataset using actor, director, and genre information. The first query identifies highly rated Science Fiction and Action movies featuring Bruce Willis. The second identifies movies featuring Uma Thurman and directed by Quentin Tarantino, ordered by runtime from shortest to longest.

```python
search_1 = df[
    df["cast"].str.contains("Bruce Willis", case=False, na=False)
    & df["genres"].str.contains("Science Fiction", case=False, na=False)
    & df["genres"].str.contains("Action", case=False, na=False)
]

search_1 = search_1.sort_values("vote_average", ascending=False)

display(search_1[
    ["title", "genres", "vote_average", "vote_count", "cast"]
])
```

```python
search_2 = df[
    df["cast"].str.contains("Uma Thurman", case=False, na=False)
    & df["director"].str.contains(
        "Quentin Tarantino", case=False, na=False
    )
]

search_2 = search_2.sort_values("runtime", ascending=True)

display(search_2[
    ["title", "director", "runtime", "vote_average", "cast"]
])
```

### 8. Most Successful Movie Franchises

I will group movies by their collection name and calculate the number of movies, total and mean production budget, total and mean revenue, and mean rating for each franchise. These metrics will help compare the financial performance and audience ratings of different franchises.

```python
franchise_df = df[df["belongs_to_collection"].notna()].copy()

franchise_summary = (
    franchise_df.groupby("belongs_to_collection")
    .agg(
        total_movies=("id", "count"),
        total_budget_musd=("budget_musd", "sum"),
        mean_budget_musd=("budget_musd", "mean"),
        total_revenue_musd=("revenue_musd", "sum"),
        mean_revenue_musd=("revenue_musd", "mean"),
        mean_rating=("vote_average", "mean"),
    )
    .sort_values("total_revenue_musd", ascending=False)
)

display(franchise_summary.round(2))
```

```python
print("\nTop 10 franchises by total movies:")
display(
    franchise_summary
    .sort_values("total_movies", ascending=False)
    .head(10)
    .round(2)
)
```

```python
print("\nTop 10 franchises by total revenue:")
display(franchise_summary.head(10).round(2))
```

### 9. Most Successful Directors

I will group movies by director and calculate the number of movies directed, total revenue, and mean rating. This will help identify directors with the highest combined box-office revenue and compare their average audience ratings.

```python
director_df = df.dropna(subset=["director"]).copy()

director_df["director"] = (
    director_df["director"].str.split(" | ", regex=False)
)

director_df = director_df.explode("director")

director_df["director"] = director_df["director"].str.strip()

director_df = director_df[
    director_df["director"].ne("")
]
```

Now calculate the director metrics:

```python
director_summary = (
    director_df.groupby("director")
    .agg(
        total_movies=("id", "nunique"),
        total_revenue_musd=("revenue_musd", "sum"),
        mean_rating=("vote_average", "mean"),
    )
    .sort_values("total_revenue_musd", ascending=False)
)

display(director_summary.round(2))
```

Show the directors with the most movies in the dataset:

```python
display(
    director_summary
    .sort_values("total_movies", ascending=False)
    .head(10)
    .round(2)
)
```
